```python
from collections import deque

def solve(start, goal):
    q = deque([(start, [])])
    visited = {start}

    while q:
        state, path = q.popleft()

        if state == goal:
            return path + [state]

        i = state.index(0)
        r, c = divmod(i, 3)

        for dr, dc in [(-1,0),(1,0),(0,-1),(0,1)]:
            nr, nc = r + dr, c + dc

            if 0 <= nr < 3 and 0 <= nc < 3:
                j = nr * 3 + nc
                new = list(state)
                new[i], new[j] = new[j], new[i]
                new = tuple(new)

                if new not in visited:
                    visited.add(new)
                    q.append((new, path))

start = (1,2,3,4,0,6,7,5,8)
goal = (1,2,3,4,5,6,7,8,0)

for s in solve(start, goal):
    print(s[:3])
    print(s[3:6])
    print(s[6:])
    print()


print("NAME :KEERTHANA T USN:1WN24CS137")

```

```python
start = (
    ("1", "2", "3"),
    ("4", "_", "6"),
    ("7", "5", "8")
)

goal = (
    ("1", "2", "3"),
    ("4", "5", "6"),
    ("7", "8", "_")
)


def dls(state, goal, depth, path):

    if state == goal:
        return path

    if depth == 0:
        return None

    for r in range(3):
        for c in range(3):
            if state[r][c] == "_":
                blank_r = r
                blank_c = c
                break
        else:
            continue
        break

    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]

    for dr, dc in moves:

        new_r = blank_r + dr
        new_c = blank_c + dc

        if 0 <= new_r < 3 and 0 <= new_c < 3:

            new_state = [list(row) for row in state]

            new_state[blank_r][blank_c], new_state[new_r][new_c] = \
                new_state[new_r][new_c], new_state[blank_r][blank_c]

            new_state = tuple(tuple(row) for row in new_state)

            if new_state not in path:

                answer = dls(
                    new_state,
                    goal,
                    depth - 1,
                    path + [new_state]
                )

                if answer is not None:
                    return answer

    return None


def iddfs(start, goal):

    for depth in range(1, 10):

        print("Searching depth:", depth)

        answer = dls(
            start,
            goal,
            depth,
            [start]
        )

        if answer is not None:
            return answer

    return None


solution = iddfs(start, goal)

if solution:

    print("\nSolution Found!")
    print("Number of moves:", len(solution) - 1)
    print()

    for step, state in enumerate(solution):

        print("Step", step)

        for row in state:
            print(" ".join(row))

        print()

else:

    print("No solution found within depth limit.")
    print("NAME :KEERTHANA T USN:1WN24CS137")
```
