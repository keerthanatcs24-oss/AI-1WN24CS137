```python

import itertools

def evaluate(expr, model):
    return eval(expr, {}, model)

def entails(kb, query, symbols):
    for values in itertools.product([True, False], repeat=len(symbols)):
        model = dict(zip(symbols, values))
        kb_true = all(evaluate(expr, model) for expr in kb)
        if kb_true and not evaluate(query, model):
            return False
    return True


if __name__ == "__main__":
    kb = [
        "(P == Q)",
        "(Q or R)",
        "(P or R)"
    ]

    symbols = ["P", "Q", "R"]

    print("P Q R | P<->Q | Q or R | P or R | KB")
    print("-" * 43)

    for values in itertools.product([True, False], repeat=3):
        model = dict(zip(symbols, values))

        p_q = evaluate("(P == Q)", model)
        q_r = evaluate("(Q or R)", model)
        p_r = evaluate("(P or R)", model)
        kb_true = all(evaluate(expr, model) for expr in kb)

        tf = lambda x: "T" if x else "F"

        print(
            tf(model["P"]), tf(model["Q"]), tf(model["R"]), "|",
            tf(p_q), "   |",
            tf(q_r), "     |",
            tf(p_r), "     |",
            tf(kb_true)
        )

    queries = {
        "Q": "Q",
        "P->R": "(not P) or R",
        "R->Q": "(not R) or Q"
    }

    print("\nEntailment Results:")
    for name, expr in queries.items():
        print(f"Does KB entail {name}? {entails(kb, expr, symbols)}")
```
