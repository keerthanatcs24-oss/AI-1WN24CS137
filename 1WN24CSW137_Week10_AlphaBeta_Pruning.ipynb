```python
import math

def alpha_beta(node, alpha, beta, maximizing):
    if isinstance(node, int):
        return node

    if maximizing:
        best = -math.inf

        for child in node:
            value = alpha_beta(child, alpha, beta, False)
            best = max(best, value)
            alpha = max(alpha, best)

            if beta <= alpha:
                print("Pruned:", node[node.index(child)+1:])
                break

        return best

    else:
        best = math.inf

        for child in node:
            value = alpha_beta(child, alpha, beta, True)
            best = min(best, value)
            beta = min(beta, best)

            if beta <= alpha:
                print("Pruned:", node[node.index(child)+1:])
                break

        return best


# Game tree
tree = [
    [[4, 3], [6, 2]],
    [[2, 1], [3, 5], [3, 1]],
    [[5, 4], [7, 5]]
]

result = alpha_beta(tree, -math.inf, math.inf, True)

print("Optimal value:", result)

print("1WN24CS137 = KEERTHANA T")
```
